## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

The starter dataset contains 30,000 pages. I used the available CTR and search-position data to check whether there is enough variation to make a CTR-opportunity question worth investigating.

The overall correlation between average search position and CTR is -0.073. This is a weak negative linear relationship, so average position alone does not explain most of the variation in CTR.

However, the position tiers show more noticeable differences in average CTR. For example, the average CTR is 0.334 for pages in the top 3 positions, compared with 0.055 for pages in the deep position tier.

These results suggest that CTR should not be judged using one overall threshold. A page's search position and other available page-level characteristics may need to be considered when identifying potential CTR opportunities.taset.

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Total pages:", len(df))

corr = df["avg_position"].corr(df["ctr"])
print(f"Correlation between avg_position and CTR: {corr:.3f}")

print("\nAverage CTR:", round(df["ctr"].mean(),3))
print("Average search position:", round(df["avg_position"].mean(),2))

visible = df[df["impressions_90d"] >= 100]

print("\nAverage CTR by position tier:")
print(
    visible.groupby("position_tier")["ctr"]
    .mean()
    .round(3)
)
```
