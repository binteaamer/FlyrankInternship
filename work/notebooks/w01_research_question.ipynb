## 2. The question: decision, action, cost of a wrong call

### Decision

Which pages should be prioritized as potential quick-win opportunities for targeted SEO optimization?

### Who acts?

An SEO specialist or content strategist reviews the prioritized pages and decides whether an optimization is appropriate.

### Action

For a page identified as a potential quick win, the specialist could:

- review the page's search position and impressions;
- review the title and meta description;
- check whether the content matches the search intent;
- review the page's content and structure;
- make a targeted SEO improvement if appropriate;
- monitor the page's ranking after the change.

The model would support prioritization. It would not automatically decide which SEO change should be made.

### Cost of a wrong recommendation

A false positive could cause the SEO team to spend time optimizing a page that has little potential for improvement.

A false negative could cause the team to miss a page that had a good opportunity for a ranking improvement.

The cost is therefore mainly wasted optimization effort or missed opportunities.

### Why data or ML can help

A page's average search position alone does not tell us whether it is a good quick-win opportunity.

Two pages can have similar positions but very different impressions, CTR, content characteristics, and engagement.

ML may be useful because it can combine several signals and learn patterns associated with pages that later achieve meaningful ranking improvements.

The goal is therefore not simply to produce a score. The goal is to create a ranked list that helps an SEO specialist decide where to spend optimization effort first.

## 3. Quick look at the data

The starter dataset contains 30,000 content items and includes several signals that are relevant to the Quick-Win question.

In particular, it contains average search position, impressions, clicks, CTR, content age, freshness, engagement, and content characteristics.

These signals allow us to identify pages that already have measurable search visibility while still having room to improve their ranking.

The following code shows some basic numbers from the starter dataset.

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Total pages:", len(df))

print("Average search position:",
      round(df["avg_position"].mean(), 2))

print("Average impressions over 90 days:",
      round(df["impressions_90d"].mean(), 2))

print("Average CTR:",
      round(df["ctr"].mean(), 3))

print("\nPosition tier counts:")
print(df["position_tier"].value_counts())
```

```python
print("\nPages with average position between 5 and 15:")

quick_win_position = df[
    (df["avg_position"] >= 5) &
    (df["avg_position"] <= 15)
]

print(len(quick_win_position))
```

### What these numbers suggest

There are 30,000 pages in the starter dataset, with an average search position of 16.34.

I identified 13,739 pages with an average position between 5 and 15. This is approximately 45.8% of the dataset. These pages are visible in search but are not yet in the top 3 positions, making them a reasonable candidate group for investigating potential ranking opportunities.

However, being in positions 5–15 does not by itself mean that a page is a quick win. The next step is to determine whether other available signals, such as impressions, CTR, content characteristics, and engagement, can help distinguish pages with stronger future improvement potential.

Therefore, the 5–15 range is an initial candidate definition, not a proven rule.
