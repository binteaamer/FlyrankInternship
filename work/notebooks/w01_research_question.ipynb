## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

To understand whether this question is worth investigating, I measured the relationship between average search position and click-through rate and summarised both variables using the starter dataset.

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Total pages:", len(df))

corr = df["avg_position"].corr(df["ctr"])
print(f"Correlation between avg_position and CTR: {corr:.3f}")

print("\nAverage CTR:", round(df["ctr"].mean(),3))
print("Average search position:", round(df["avg_position"].mean(),2))

visible = df[df["impressions_90d"] >= 100]

print("\nAverage CTR by position tier:")
print(
    visible.groupby("position_tier")["ctr"]
    .mean()
    .round(3)
)
```
