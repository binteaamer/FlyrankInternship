## 1. My lane as an ML task (type)

**Task type: Ranking / scoring**

My lane is a Quick-Win Opportunity Detection problem. The goal is to rank visible pages by how promising they are for targeted SEO optimisation.

The output will be a priority score or ranked list rather than only a yes/no prediction. Pages can have different levels of opportunity, so ranking them allows an SEO specialist to decide which pages should be reviewed first.

The ranking can use multiple signals such as current search position, impressions, CTR, recent performance, query coverage, and other page-level characteristics.

The decision is not simply "which pages rank at position 5?" Instead, the goal is to identify pages that have a combination of visibility, ranking proximity, and other signals that may indicate an opportunity for improvement.

## 2. Target or proxy

The preferred target is an observed future ranking improvement.

For example, for pages that are currently in a reasonable opportunity range, we can measure whether their average search position improves meaningfully during a later time window.

A possible observed outcome is whether a page moves substantially closer to the top of search results during the following 30 days.

This target will be created from future performance data rather than from the same rules used to select Quick-Win candidates. This is important because a model should learn from an observed outcome rather than simply reproduce a hand-written rule.

At this stage, the exact improvement threshold is provisional and will be chosen after examining the distribution of future position changes in the full warehouse.

If the future ranking outcome is too noisy or too rare, I will use a clearly defined observed proxy such as meaningful improvement in average search position or movement into a higher position tier.

## 3. Success metric

The main success metric will be **Precision@K**, with K initially set to 20.

Precision@20 will measure the proportion of the top 20 pages in the model's Quick-Win queue that actually achieve the defined future ranking improvement.

For example, if 8 of the 20 highest-ranked pages later show the required improvement, Precision@20 would be 0.40.

This metric matches the business decision because an SEO team has limited time and would normally review a small number of pages first. A useful ranking should therefore put a high proportion of genuinely improving pages near the top of the queue.

The exact target definition will be finalized before model evaluation so that the success metric is not changed after seeing the results.

## 5. Why ML beats a fixed rule here

A simple rule such as "select every page between positions 5 and 15" would be easy to implement, but it treats all pages in that range as equally promising.

In reality, pages with similar positions can have very different levels of impressions, CTR, query coverage, recent momentum, and search-performance patterns.

For example, a page at position 7 with strong visibility and many relevant queries may represent a different opportunity from another page at position 7 with very little demand or worsening performance.

A ranking model can combine several signals and learn which combinations are associated with the observed future outcome. This makes the output more useful for prioritisation than a single fixed threshold.

However, ML only earns its place if it improves ranking performance against a simple baseline. I will therefore compare the model against a fixed-rule baseline rather than assuming that ML is automatically better.
